```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("AI_Learner_EV_Dataset_Pro.csv")
```

```python
# clean data
df = df.drop_duplicates()
df.columns = df.columns.str.lower().str.replace(" ", "_")
df = df.dropna()

print("Data Loaded Successfully\n")
```

```python
num_data = df.select_dtypes(include=np.number)

print("Average Values:\n", num_data.mean())
print("\nStandard Deviation:\n", num_data.std())
```

```python
plt.hist(df['priceeuro'], bins=20)
plt.title("EV Price Distribution")
plt.xlabel("Price (Euro)")
plt.ylabel("Number of Vehicles")
plt.show()

print("Insight: EV prices are widely spread, indicating multiple market segments.\n")
```

```python
plt.hist(df['range_km'], bins=20)
plt.title("Range Distribution")
plt.xlabel("Range (km)")
plt.ylabel("Number of Vehicles")
plt.show()

print("Insight: Range varies significantly and is a key differentiator.\n")
```

```python
plt.scatter(df['range_km'], df['priceeuro'])
plt.title("Price vs Range (Key Relationship)")
plt.xlabel("Range (km)")
plt.ylabel("Price (Euro)")
plt.grid(True)
plt.show()
```

```python
df['brand'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 EV Brands")
plt.xlabel("Brand")
plt.ylabel("Number of Models")
plt.show()

print("Insight: A few brands dominate the EV market.\n")
```

```python
df['powertrain'].value_counts().plot(kind='bar')
plt.title("Powertrain Types")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

print("Insight: Certain powertrain types are more widely used.\n")
```

```python
df['segment'].value_counts().plot(kind='bar')
plt.title("Vehicle Segment Distribution")
plt.xlabel("Segment")
plt.ylabel("Count")
plt.show()

print("Insight: EVs are concentrated in specific segments.\n")
```

```python
plt.hist(df['topspeed_kmh'], bins=20)
plt.title("Top Speed Distribution")
plt.xlabel("Top Speed (km/h)")
plt.ylabel("Count")
plt.show()

print("Insight: Most EVs have similar top speeds, showing consistency.\n")
```

```python
plt.hist(df['accelsec'], bins=20)
plt.title("Acceleration Distribution (0-100 km/h)")
plt.xlabel("Seconds")
plt.ylabel("Count")
plt.show()

print("Insight: Acceleration is relatively consistent across vehicles.\n")
```

```python
plt.hist(df['efficiency_whkm'], bins=20)
plt.title("Efficiency Distribution")
plt.xlabel("Wh/km")
plt.ylabel("Count")
plt.show()

print("Insight: Efficiency varies across models, indicating optimization differences.\n")
```

```python
plt.hist(df['price_per_range'], bins=20)
plt.title("Price per Range Distribution")
plt.xlabel("Price per km")
plt.ylabel("Count")
plt.show()

print("Insight: Some vehicles offer better value for money than others.\n")
```

```python
corr = np.corrcoef(num_data.T)

plt.imshow(corr)
plt.colorbar()
plt.title("Correlation Heatmap")

plt.xticks(range(len(num_data.columns)), num_data.columns, rotation=90)
plt.yticks(range(len(num_data.columns)), num_data.columns)

plt.show()

print("Insight: Price and range are strongly related.\n")
```

```python
avg_price_brand = df.groupby('brand')['priceeuro'].mean().sort_values(ascending=False).head(10)

avg_price_brand.plot(kind='bar')
plt.title("Average Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price (Euro)")
plt.show()

print("Insight: Premium brands command significantly higher prices, indicating strong brand positioning and perceived value.")
```

```python
plt.boxplot(df['priceeuro'])
plt.title("Price Outliers")
plt.show()

print("Insight: Presence of premium vehicles increases price spread.\n")
```

```python
plt.scatter(df['range_km'], df['efficiency_whkm'])
plt.title("Range vs Efficiency")
plt.xlabel("Range (km)")
plt.ylabel("Efficiency (Wh/km)")
plt.show()

print("Insight: Vehicles with higher range often require more energy, highlighting a trade-off between performance and efficiency.")
```

```python
plt.boxplot(df['range_km'])
plt.title("Range Outliers")
plt.show()

print("Insight: Some EVs offer exceptionally high range.\n")
```

```python
df.groupby('segment')['priceeuro'].mean().plot(kind='bar')
plt.title("Average Price by Segment")
plt.xlabel("Segment")
plt.ylabel("Average Price")
plt.show()

print("Insight: Certain segments are positioned as premium, while others target budget consumers.")
```

```python
print("FINAL INSIGHTS")
print("1. EV market is highly segmented with wide price variation.")
print("2. Range is the strongest driver of pricing and competitiveness.")
print("3. Premium brands dominate higher price segments.")
print("4. Performance metrics like speed and acceleration are relatively stable.")
print("5. Efficiency and range show a trade-off relationship.")
print("6. Product positioning varies significantly across segments.\n")
```

```python
print("CONCLUSION")
print("The EV market is driven by range, pricing, and brand positioning.")
print("Companies that improve range while maintaining cost efficiency will gain a competitive advantage.")
print("Strategic segmentation and value positioning are key to success.\n")
```
